```python
# ---------------- Task 1: List Manipulation ----------------
print("\n--- Task 1: List Manipulation ---")

# Create a list
fruits = ["apple", "banana", "cherry"]
print("Original List:", fruits)

# Add an item
fruits.append("mango")
print("After Adding 'mango':", fruits)

# Remove an item
fruits.remove("banana")
print("After Removing 'banana':", fruits)

# Sort the list
fruits.sort()
print("After Sorting:", fruits)

# Reverse the list
fruits.reverse()
print("After Reversing:", fruits)
```

```python
# ---------------- Task 2: String Manipulation ----------------
print("\n--- Task 2: String Manipulation ---")

text1 = "Hello"
text2 = "World"

# Concatenation
combined = text1 + " " + text2
print("Concatenation:", combined)

# Slicing
print("Sliced (0 to 4):", combined[0:5])

# Reversing
print("Reversed String:", combined[::-1])
```

```python
# ---------------- Task 3: Array Manipulation ----------------
print("\n--- Task 3: Array Manipulation ---")

from array import array

# Create an integer array
numbers = array('i', [10, 20, 30, 40])
print("Original Array:", numbers.tolist())

# Append a new value
numbers.append(50)
print("After Append:", numbers.tolist())

# Remove a value
numbers.remove(20)
print("After Remove:", numbers.tolist())

# Access elements
print("First Element:", numbers[0])
print("Last Element:", numbers[-1])

```

```python
# ---------------- Task 4: Dictionary (Student Records) ----------------
print("\n--- Task 4: Dictionary (Student Records) ---")

# Create a dictionary
student = {
    "name": "Ali",
    "id": 101,
    "grade": "A"
}
print("Original Record:", student)

# Create (Add new key)
student["age"] = 20
print("After Adding Age:", student)

# Read
print("Student Name:", student["name"])

# Update
student["grade"] = "A+"
print("After Updating Grade:", student)

# Delete
del student["age"]
print("After Deleting Age:", student)
```
