```python
# Event Registration System

participants = {}

while True:
    print("\n--- Tech Conference Registration ---")
    name = input("Enter participant name (or 'q' to quit): ")

    if name.lower() == 'q':
        break

    email = input("Enter email address: ")

    try:
        if email in participants.values():
            raise ValueError("This email is already registered! Please enter a unique email.")
        participants[name] = email
        print(f"{name} has been successfully registered!")
    except ValueError as e:
        print(e)

print("\n--- Registration Complete ---")
print(f"Total participants registered: {len(participants)}")

print("\nList of Participants:")
for i, (name, email) in enumerate(participants.items(), start=1):
    print(f"{i}. {name} - {email}")
```

```python
# Text File Analyzer

filename = input("Enter the file name (with .txt extension): ")

try:
    with open(filename, "r") as file:
        content = file.read()
        char_count = len(content)
        print(f"\nTotal number of characters in '{filename}': {char_count}")
except FileNotFoundError:
    print("File not found. Please make sure the file exists.")
```
